# **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
# Save the cleaned dataset after handling missing values and duplicates.
df = df.drop_duplicates().copy()
df['Released_Year'] = pd.to_numeric(df['Released_Year'], errors='coerce').astype('Int64')
df['Runtime'] = pd.to_numeric(df['Runtime'].astype(str).str.replace(' min', '', regex=False), errors='coerce').astype('Int64')
df['Gross'] = pd.to_numeric(df['Gross'], errors='coerce')
df['Gross'] = df['Gross'].fillna(df['Gross'].median())
df.to_csv('cleaned_movies.csv', index=False)
print('Saved as cleaned_movies.csv')
```

# **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
genre_rows = df[df['Genre'].str.contains(r'Drama|Action', case=False, na=False)].copy()
genre_rows['Genre_Group'] = genre_rows['Genre'].apply(lambda x: 'Drama' if 'Drama' in str(x) else 'Action')
plot_data = [genre_rows.loc[genre_rows['Genre_Group'] == g, 'Gross'].dropna() for g in ['Drama', 'Action']]
plt.figure(figsize=(8, 5))
plt.boxplot(plot_data, tick_labels=['Drama', 'Action'])
plt.title('Gross Revenue: Drama vs Action')
plt.ylabel('Gross Revenue')
plt.tight_layout()
plt.show()
```

# **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = df.nlargest(5, 'IMDB_Rating').copy()
top5['IMDB_Rating_100'] = top5['IMDB_Rating'] * 10
comparison = top5.set_index('Series_Title')[['Meta_score', 'IMDB_Rating_100']]
comparison.plot(kind='bar', figsize=(10, 6))
plt.title('Top 5 Rated Movies: Meta Score vs IMDb Rating (scaled to 100)')
plt.xlabel('Movie')
plt.ylabel('Score')
plt.xticks(rotation=45, ha='right')
plt.legend(['Meta Score', 'IMDb Rating × 10'])
plt.tight_layout()
plt.show()
```
